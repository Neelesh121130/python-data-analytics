```python
# project building a simple calculator for the simple calculation it is a very simple project that everyone shoud easily build now see the project
#first take the input from the user
num1=int(input("Enter first number: "))
num2=int(input("Enter second number: "))
#user should enter the opertion from the given operations
operation=input("Enter operation (+, -, *, /): ")
#now using loop for the correct answer
if operation=='+':
    print(num1+num2)
elif operation=='-':
    print(num1-num2)
elif operation=='*':
    print(num1*num2)

elif operation=='/':
    if num2!=0:
        print(num1/num2)
    else:
        print("Error: Division by zero is not allowed.")
else:
    print("Invalid operation. Please enter one of +, -, *, /.")
```
